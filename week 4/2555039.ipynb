```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
!pip install requests beautifulsoup4 pandas scikit-learn
```

```python
# 1. Scraping Data
quotes = []
print("Scraping quotes...")

for page in range(1, 11):
    url = f"http://quotes.toscrape.com/page/{page}/"
    response = requests.get(url)
    response.encoding = "utf-8"

    soup = BeautifulSoup(response.text, "html.parser")
    quote_elements = soup.select(".quote .text")

    for q in quote_elements:
        quotes.append(q.text)
```

```python
# 2. Creating DataFrame
df = pd.DataFrame(quotes, columns=["quote"])
```

```python
# 3. Text Analysis (Word Frequency)
# We use stop_words='english' to ignore common words like 'the', 'is', 'and'
vectorizer = CountVectorizer(stop_words='english')
X = vectorizer.fit_transform(df["quote"])

# Calculate frequencies
word_counts = X.toarray().sum(axis=0)
words = vectorizer.get_feature_names_out()

freq_df = pd.DataFrame({"word": words, "count": word_counts})
top5 = freq_df.sort_values(by="count", ascending=False).head(5)
```

```python
# 4. Binary Encoding (One-Hot Representation)
vectorizer_bin = CountVectorizer(stop_words='english', binary=True)
X_bin = vectorizer_bin.fit_transform(df["quote"])
df_encoded = pd.DataFrame(X_bin.toarray(), columns=vectorizer_bin.get_feature_names_out())
```

```python
# 5. Output Results
print("-" * 30)
print(f"Total quotes collected: {len(df)}")
print("\nTOP 5 MOST FREQUENT WORDS:")
print(top5.to_string(index=False))
print("-" * 30)
print("\nPreview of Encoded DataFrame (First 5 rows):")
print(df_encoded.head())
```
